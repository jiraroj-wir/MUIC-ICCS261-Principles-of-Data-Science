### Story structure (fill in all six parts)

1. **Hook** (1–2 sentences): why should the audience care?
2. **Big idea** (one sentence): your main message. It must contain a number.
3. **Hero chart**: **one** chart that carries the big idea. Up to **two** supporting charts or tables.
4. **"So what?"**: what does it mean? Who was affected?
5. **Caveats** (at least 2): what the data *cannot* tell us. Consider:

6. **Next question**: one question you would test formally later (Module 2 Topics 3–4: sampling and significance testing).

### ✔️ Hero-chart checklist
- [ ] The title **states the finding** (e.g., "Third-class men had the lowest survival: 14%"), not just the variables.

- [ ] Group sizes (*n*) are visible in the chart or caption.
- [ ] Bars are sorted or ordered meaningfully. No pie chart with more than 3 slices.
- [ ] Colour highlights **one** thing; everything else is grey or muted.
- [ ] Built with Seaborn, then refined with Matplotlib (slides: *"Many users combine Seaborn… with Matplotlib for final, detailed refinements"*).
