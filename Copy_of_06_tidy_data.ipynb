#### 1.1 What is tidy data?

Hadley Wickham proposed the term **TIDY DATA**.

>A huge amount of effort is spent cleaning data to get it ready for analysis, but there
has been little research on how to make data cleaning as easy and effective as possible.
This paper tackles a small, but important, component of data cleaning: data tidying.
Tidy datasets are easy to manipulate, model and visualise, and have a specific structure:
each variable is a column, each observation is a row, and each type of observational unit
is a table. This framework makes it easy to tidy messy datasets because only a small
set of tools are needed to deal with a wide range of un-tidy datasets. This structure
also makes it easier to develop tidy tools for data analysis, tools that both input and
output tidy datasets. The advantages of a consistent data structure and matching tools
are demonstrated with a case study free from mundane data manipulation chores. <br>-- Hadley Wickham https://www.jstatsoft.org/article/view/v059i10

The rules for data to be in this form are:

1. Each variable must have its own column.
2. Each observation must have its own row.
3. Each value must have its own cell.

A. Drop the total column, then melt.

The column `new_sp` is the **total** of all the other columns, so it is a different kind of observation (an aggregate). If we keep it, a row labelled "total" would sit next to the rows it summarizes, and any later sum would count the same cases twice.

```python
# Remove the aggregate column before reshaping.
# errors='ignore' means: do nothing (instead of failing) if the column is already gone.
df_tb_detail = df_tb.drop(columns=['new_sp'], errors='ignore')

# Melt the 20 sex-and-age columns into rows.
df_tb_m = pd.melt(df_tb_detail,
                  id_vars=["iso2", "year"],    # identifiers kept on every row
                  value_name="cases",          # the numbers go here
                  var_name="sex_and_age"       # the old column names (e.g. 'new_sp_m014') go here
                  )

# info() prints its report directly, so it does not need display().
df_tb_m.info()
df_tb_m
```

```python
# sample(10) shows 10 random rows; useful for spot-checking a large table.
df_tb_m.sample(10)
```

B. Split the <code>sex_and_age</code> column into two variables: sex and age.<br>If you would like to know more about regular expressions, check out [How to write Regular Expressions?](https://www.geeksforgeeks.org/write-regular-expressions/)

The codes after `new_sp_` follow this pattern: one letter for sex, then an age code.

| Age code | Meaning |
|---|---|
| `04` | 0-4 |
| `514` | 5-14 |
| `014` | 0-14 |
| `1524`, `2534`, ..., `5564` | 15-24, 25-34, ..., 55-64 |
| `65` | 65 and over |
| `u` | unknown |

Because the age codes have different lengths (2, 3, or 4 characters), the digits cannot be split by position alone. We first *extract* the sex letter and the age code with a regular expression, then translate each age code with a lookup table.

```python
import numpy as np

# Regular expression, piece by piece:
#   ^new_sp_      the text must start with 'new_sp_'
#   (?P<sex>[mf]) capture one letter, m or f, and call the group 'sex'
#   (?P<age_code>\d+|u)  capture either one or more digits, or the letter u, as 'age_code'
#   $             nothing may follow (so no partial matches)
pattern = r'^new_sp_(?P<sex>[mf])(?P<age_code>\d+|u)$'

# str.extract applies the pattern to every row and returns one column per named group.
tmp_df = df_tb_m['sex_and_age'].str.extract(pattern)

# Lookup table: age code -> readable age range.
AGE_LABELS = {
    '04': '0-4',
    '514': '5-14',
    '014': '0-14',
    '1524': '15-24',
    '2534': '25-34',
    '3544': '35-44',
    '4554': '45-54',
    '5564': '55-64',
    '65': '65+',
    'u': 'unknown',
}

# map() replaces each code with its label; any code not in the table becomes NaN.
tmp_df['age'] = tmp_df['age_code'].map(AGE_LABELS)

# Safety check: stop with an error if any column name did not match the pattern
# or used an age code we did not expect. Silent NaNs are how parsing bugs hide.
unmatched = df_tb_m.loc[tmp_df['age'].isna(), 'sex_and_age'].unique()
assert len(unmatched) == 0, f"Unexpected column codes: {unmatched}"

display(tmp_df.sample(10))
```

C. Add the new columns to the melted table and drop the old combined column.

```python
# Copy the parsed columns into df_tb_m.
# Both tables share the same row index, so pandas lines up the rows automatically.
df_tb_m['sex'] = tmp_df['sex']
df_tb_m['age'] = tmp_df['age']

# 'sex_and_age' is now redundant: its information lives in 'sex' and 'age'.
df_tb_m = df_tb_m.drop(columns=['sex_and_age'])

# Put the columns in a sensible order: identifiers first, measurement last.
df_tb_m = df_tb_m[['iso2', 'year', 'sex', 'age', 'cases']]

df_tb_m
```

Let's load the data file, and quickly clean it like we did before.

```python
import pandas as pd
import numpy as np
import urllib.request   # for downloading a file from a URL
import re               # regular expressions, for pulling numbers out of text


data = []   # will hold one dictionary per line of the file

url = 'https://drive.google.com/uc?export=download&id=1d9b_at4SJMlgtzQ04isPjUVzGEm7hYsl'

# Download the raw file. The 'with' block closes the connection automatically.
with urllib.request.urlopen(url) as f:
    content_bytes = f.read()                      # raw bytes from the web
    text_content = content_bytes.decode('utf-8')  # bytes -> text
    lines = text_content.splitlines()             # text -> list of lines

    # Show the first and last 10 lines so we can see what we are dealing with.
    _ = [print(line) for line in lines[:10]]
    print('...')
    _ = [print(line) for line in lines[-10:]]

    for line in lines:
        if not line.strip():    # skip empty lines
            continue

        # 1. Parse metadata. The first 21 characters have fixed widths:
        #    characters 0-10 station ID | 11-14 year | 15-16 month | 17-20 element (e.g. TMAX)
        station_id = line[:11]
        year = int(line[11:15])
        month = int(line[15:17])
        element = line[17:21]

        # 2. Parse data values (the rest of the line).
        # Pull out every whole number, including -9999 (the "missing" code).
        # Assumption: the flag characters between values are letters or blanks.
        # (True for this file: every line yields exactly 31 numbers.)
        # If a flag were a digit, this would pick up extra numbers and shift the days,
        # so re-check this assumption before reusing the code on another GHCN file.
        raw_data = line[21:]
        daily_values = re.findall(r'-?\d+', raw_data)

        # 3. Store the line as a dictionary: identifiers first...
        row = {
            'ID': station_id,
            'Year': year,
            'Month': month,
            'Element': element
        }

        # ...then one key per day: Day_1, Day_2, ..., Day_31.
        for i, value in enumerate(daily_values):
            row[f'Day_{i+1}'] = value

        data.append(row)

# 4. Create the initial DataFrame: still WIDE (one column per day).
df = pd.DataFrame(data)
```

```python
# Look at the wide table BEFORE fixing it: days are spread across 31 columns.
df.head()
```

```python
# 5. Transform from wide (Day_1 ... Day_31) to long (one row per day).
# This creates a 'Day_Col' column (old column names) and a 'Value' column (the readings).
id_vars = ['ID', 'Year', 'Month', 'Element']
df_long = df.melt(id_vars=id_vars, var_name='Day_Col', value_name='Value')

# Extract the day number from strings such as 'Day_1', 'Day_2', ... and store it as an integer.
df_long['Day'] = df_long['Day_Col'].str.extract(r'(\d+)', expand=False).astype(int)

# 6. Create the 'Date' column from Year, Month, Day.
# errors='coerce' turns impossible dates (e.g. February 30th) into NaT ("not a time").
df_long['Date'] = pd.to_datetime(
    df_long[['Year', 'Month', 'Day']],
    errors='coerce'
)

# Drop rows whose date is invalid (NaT). Missing readings (-9999) are handled later.
df_long = df_long.dropna(subset=['Date'])

# Store the date as text in yyyy-mm-dd format.
df_long['Date'] = df_long['Date'].dt.strftime('%Y-%m-%d')

# 7. Final cleanup
# Keep only the columns we need, in a readable order.
# Note: TMAX, TMIN and PRCP values are still mixed together in one 'Value' column.
final_df = df_long[['ID', 'Date', 'Element', 'Value']]

# Sort for readability and renumber the rows from 0.
df_wt = final_df.sort_values(by=['ID', 'Date', 'Element']).reset_index(drop=True)
```

```python
display(df_wt)
display(df_wt.sample(10))
```

The last two things to do are
- to replace `-9999` with `numpy.nan` (GHCN uses -9999 to mean "missing").
- to convert the units: GHCN stores temperatures in tenths of °C and precipitation in tenths of mm, so we divide by 10.

```python
# The values were read as text; convert them to numbers.
# errors='coerce' turns anything that is not a number into NaN.
df_wt['Value'] = pd.to_numeric(df_wt['Value'], errors='coerce')

# Replace the sentinel value -9999 with NaN, in the 'Value' column only.
df_wt['Value'] = df_wt['Value'].replace(-9999, np.nan)

# Convert tenths to whole units. Dividing a column applies to every row at once,
# and NaN / 10 stays NaN, so no special handling is needed.
df_wt['Value'] = df_wt['Value'] / 10

df_wt
```

Our work is not done yet. As suggested in the paper, we need one more step to put each observation in its own row: the `Element` column holds variable *names* (TMAX, TMIN, PRCP), so it should be spread into columns.

<img src="https://i.ibb.co/kQHMTB7/final-weather.png" width=60%>

```python
# pivot is the opposite of melt: values in 'Element' become new column names.
# One row per (ID, Date); one column per element.
# pivot raises an error if any (ID, Date, Element) appears twice, which is a useful check.
# (pivot_table would silently average duplicates instead.)
df_wt_pivoted = df_wt.pivot(
    index=['ID', 'Date'],
    columns='Element',
    values='Value'
).reset_index()   # turn ID and Date back into ordinary columns

df_wt_pivoted.columns.name = None   # remove the leftover 'Element' label above the column names

# Display the tidied DataFrame
display(df_wt_pivoted)
display(df_wt_pivoted.sample(10))
```

#### 1.4.5 One type in multiple tables
- Data values about a single type of observational unit spread out over multiple tables or files
- For example, store the weather data into three files--`min_temp.csv`, `max_temp.csv`, and `prcp.csv`.
- How to fix this: merge datasets

First, let's create the problem on purpose by splitting our tidy weather table into three.

```python
# Each table keeps the identifiers (ID, Date) plus ONE measured variable.
df_min_temp = df_wt_pivoted[['ID', 'Date', 'TMIN']]
df_max_temp = df_wt_pivoted[['ID', 'Date', 'TMAX']]
df_prcp = df_wt_pivoted[['ID', 'Date', 'PRCP']]

print("df_min_temp head:")
display(df_min_temp.head())
print("\ndf_max_temp head:")
display(df_max_temp.head())
print("\ndf_prcp head:")
display(df_prcp.head())
```

Now fix it by merging the tables back together on the columns they share.

```python
# merge joins two tables row by row wherever the key columns match.
# on=['ID', 'Date']   : the columns that identify the same observation in both tables
# how='outer'         : keep a date even if it appears in only one table
# validate='one_to_one': stop with an error if a key appears more than once in either table
df_merged = (df_min_temp
             .merge(df_max_temp, on=['ID', 'Date'], how='outer', validate='one_to_one')
             .merge(df_prcp,     on=['ID', 'Date'], how='outer', validate='one_to_one'))

# Check: the merged table should have the same shape as the table we started from.
print(df_merged.shape, df_wt_pivoted.shape)
df_merged.head()
```
