Most of the time in this class, we'll reading data from files and accessing those data rather than creating our own datasets from unstructured or semi-structured data. Thus, this notebook focuses on the intricacy of read operations in `pandas`. As mentioned before in class, some methods are only available via `Series` and some only via 'DataFrame`. If our operation does not result in the correct type, we cannot apply the property of method that we need to work efficiently. (Of course, there is a way to go around things, but using a trustable, available function will save you time and increase the accuracy of your program.)

```python
import pandas as pd
```

```python
df = pd.DataFrame({
    'Gold': [15, 12, 10],
    'Silver': [11, 9, 5] ,
    'Bronze': [6, 8, 12],
    'Continent': ['Europe', 'Asia', 'Asia']
    },
    ['France','China','Thailand'])
df
```

Below are some examples to demonstrate some property and method that are available only to `pandas Series`.

```python
df.is_monotonic_increasing
```

```python
df['Gold'].is_monotonic_increasing
```

```python
df.pct_change()
```

```python
df['Gold'].pct_change()
```
