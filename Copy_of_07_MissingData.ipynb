## 1. The Problem of Missing Data

- Reduced statistical power
    - e.g., *Do we have enough cases left to detect the effect of interest?*
- Bias in the estimation of parameters
    - e.g., *Should we attribute differences in the coefficients to changes in the model or to changes in the subsample?*
- Reduced representativeness of the samples
    - e.g., *Do the estimated coefficients generalize to the study population?*
- Complicating the analysis of the study

All or some of these effects can lead to invalid conclusions/modes from the data. For example,

```python
import numpy as np

data = np.array([4, 5, np.nan])
data.mean()
```

## 2. Types of Missing Data

- **Missing completely at random (MCAR)**
  - Unrelated to both<br>a) the specific value that should have been obtained<br>**and**<br>b) the set of observed responses
    - No systematic differences exist between participants with missing data and those with complete data
  - Samples are lost accidentally or due to technical problems
  - *Pro*: Power could be lost but estimated parameters are unbiased.
  - *Con*: Don't be too fast to assume this case.
- **Missing at random (MAR)**
  - Systematically related to the observed but not the unobserved data
    - The probability that a value is missing depends on observed values of other variables
      - not on the other missing values of the target variable.
  - Example: the income level of a client can be estimated from other variables like the client’s profession, experience and qualification
- **Missing not at random (MNAR)**
  - The data doesn't fit the creiteria of MCAR and MAR
  - Example: Citizens of a country opt not participate in a survey

#### Cautions
Imputing with a single value, such as a mean:
- alters summary statistics
- changes the distribution
- inflates the presence of a specific value


Imputation with a regressor/ML model:
- sensitive to the choice of *k*
    - selecting an inappropriate value for *k* may lead to either over-smoothing (too generalized) or overfitting (too specific) imputations)
- highly computational
    - can be time-consuming for larger datasets to calculate the distance between two data points
- Can handle categorical data only when encoding it to numbers
- impact of outliers (too many outliers in the data may lead to wrong imputations)

### 3.1 Deletion
- Easiest and often the default choice, but considered to be the worst
- Assuming MCAR
- 2 main methods
  - **Listwise deletion** (aka *complete-case analysis*)
    - Deleting all observations that have missing values
    - *Con*: it may introduce serious bias especially when there are a large number of missing values and if the original data set is too small
  - **Pairwise deletion** (aka *available case analysis*)
    - Only observations with high levels of missing data are deleted
    - Choose this when the overall sample size is small or when the number of missing data observations are large

### 3.2 Single Imputation
- Fill in missing value with good guesses
- Common techniques
  - **Central of Tendency**
    - Mean, median, or mode
    - Typically used for small data samples especially in surveys
      - Not used for rigorous multivariate data systems
    - *Pro*: Simple
    - *Con*: The results may be distorted due to unevenness in the sample distribution.
  - k-nearest neighor: The closest observation as predefined by some certain characteristics to the missing value is used to donate the value.
  - **Hot-deck and cold-deck**
    - The most widely used in survey research
    - A **deck** is an N-way crosstab of 3-5 demographic variables
    - A missing value is replaced with the most similar value of that variable from the observation with similar characteristics
      - Example: imputed from the last observation in the dataset
    - Cold-deck: Use donated values from another silimar source
  - **Regression**
    - Uses complete cases to determine the realtionship between the independent variables and the dependent variable to impute
    - *Con*: Absence of an error term → biased<br>
      Data in nature do not follow one pattern.

## 4. Handle Missing Data with pandas

### 4.1 Missing Data Convention

1. **Masking**: A global indication of missing values, such as
   - An entirely separate Boolean array
   - One bit in the data representation to locally indicate the null status of a value
   - <span style="color:red; font-weight:bold;">Con</span>: Allocation of an additional Boolean array, which adds overhead in both storage and computation
2. **Sentinel value**: A value that indicates a missing entry, such as
   - Some data-specific convention, such as indicating a missing integer value with -9999
   - Some rare bit pattern
   - A more global convention, such as **NaN** (Not a Number), which is part of the IEEE floating-point specification
   - <span style="color:red; font-weight:bold;">Cons</span>:
     - Reducing the range of valid values that can be represented
     - May requiring extra (often non-optimized) logic in CPU and GPU arithmetic
     - Common special values like NaN are not available for all data types.

Here is an example of how Linear Interpolation calculates values to fill missing gaps compared to other common filling methods.

```python
# Creating the dataset illustrated above
df = pd.DataFrame({'Original': [2.0, np.nan, np.nan, 11.0]})

df['Forward_Fill'] = df['Original'].ffill()
df['Backward_Fill'] = df['Original'].bfill()
df['Interpolated'] = df['Original'].interpolate(method='linear')

print(df)
```

```python
# Gemini generated code

import matplotlib.pyplot as plt
# import numpy as np
# import pandas as pd

# 1. Create the dataset with missing (NaN) values
df = pd.DataFrame({"Original": [2.0, np.nan, np.nan, 11.0]})

# 2. Apply handling methods
df["Forward Fill (ffill)"] = df["Original"].ffill()
df["Linear Interpolation"] = df["Original"].interpolate(method="linear")

# 3. Plot the data
plt.figure(figsize=(8, 5))

# Plot the Forward Fill path (using a stepped or solid style for visibility)
plt.plot(
    df.index,
    df["Forward Fill (ffill)"],
    label="Forward Fill (ffill)",
    color="orange",
    linestyle="--",
    marker="s",
    alpha=0.7,
)

# Plot the Linear Interpolation path
plt.plot(
    df.index,
    df["Linear Interpolation"],
    label="Linear Interpolation",
    color="green",
    linestyle="-",
    marker="o",
    linewidth=2,
)

# Overlay valid original data points to highlight where the missing gaps were
valid_data = df["Original"].dropna()
plt.scatter(
    valid_data.index,
    valid_data,
    color="red",
    s=150,
    zorder=5,
    label="Original Valid Data",
)

# Highlight the interpolated points specifically
plt.annotate(
    "Interpolated (1, 5.0)",
    xy=(1, 5.0),
    xytext=(0.5, 6.5),
    arrowprops=dict(arrowstyle="->", color="black"),
)
plt.annotate(
    "Interpolated (2, 8.0)",
    xy=(2, 8.0),
    xytext=(2.1, 6.5),
    arrowprops=dict(arrowstyle="->", color="black"),
)

# Formatting the chart
plt.title("Pandas Missing Data Handling Comparison", fontsize=14, pad=15)
plt.xlabel("Index", fontsize=12)
plt.ylabel("Values", fontsize=12)
plt.xticks([0, 1, 2, 3])
plt.grid(True, linestyle=":", alpha=0.6)
plt.legend(loc="upper left")

# Display the plot
plt.tight_layout()
plt.show()
```

## 4.4 Filling Missing Data with scikit-learn

### 4.4.1 Single Imputation
- Univariate: `SimpleImputer` class
    - For univariate feature imputation
    - 'most_frequent' or 'constant' strategy
    - default = 'mean'
- Multivariate: `KNNImputer` class
    - Nearest neighbors imputation
    - A hot deck imputation technique

```python
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer
```

#### SimpleImputer class

Impute numercial features.

```python
df_nums = pd.DataFrame([[1, 2, np.nan],
                        [3, 4, 3],
                        [np.nan, 6, 5],
                        [8, 8, 7]],
                       columns=list('ABC'))
df_nums
```

```python
imp = SimpleImputer()
df_nums_new = pd.DataFrame(imp.fit_transform(df_nums))
df_nums_new
```

Impute categorial features.

```python
df_cat = pd.DataFrame([["a", "x"],
                   [np.nan, "y"],
                   ["a", np.nan],
                   ["b", "y"]], dtype="category")
imp = SimpleImputer(strategy="most_frequent")
imp.fit_transform(df_cat)
```

### 4.5 Evaluation of Imputation Methods

- Can be varied and beyond the scope for now
- Hypothesis testing in Module 2 will give you the basic ideas, and more details will be presented in Module 3's accuracy testing.
- Roughly all summary statistics (e.g., mean, standard deviation) should remain close to those of the orignal dataset. However, this depends on the objective of the cleaning process.

A simple way to try your target imputation method ([ref](https://datascience.stackexchange.com/questions/10643/how-to-measure-the-performance-of-an-imputation-technique)):
1. Take partial data without missing values from your dataset.
2. Manually delete data by trying to reproduce the overall missing data pattern. So we have the "orginal" dataset to be compared with the "imputed" dataset.
3. Do imputation on this mock dataset with the method you would also use for the overall dataset.
4. Compare your imputed values to the actual values. This gives a good estimate how far you are off for the original dataset.
