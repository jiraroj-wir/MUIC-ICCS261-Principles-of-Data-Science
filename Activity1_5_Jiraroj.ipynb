Write code to perform the following subsetting tasks on the DataFrame object `df` declared at the beginning of this notebook with numerical indices and column labels. Subsetting implies that you cannot create a new object from scratch. Using labels implies that you cannot use numerical indices.

## Part I: Getting a Specific Data Type

Use only labels indices/headers in this section. You may use `loc` and other pandas methods, but no numerical indices.

### 1.1 Access 8 as a literal.

```python
df.iloc[1, 2]
```

Access (2, 3) entry on the `dataframe`. We get a scalar in this case. 

```python
type(df.iloc[1, 2])
```

### 1.2 Access 8 as a Series object and maintain its labels.

```python
df.loc['China', ['Bronze']]
```

I usually think of it as, which ever dimension is written as a list, `['Bronze']` (our column) in this case, got to keep it's dimension, else it's dimension will be lost. 

So, in this case, we're left with (1, ), which is a `Series`. 

A `Series` also contain index labels, so that's how the label is present. 

```python
type(df.loc['China', ['Bronze']])
```

### 1.3 Access 8 as a DataFrame object and maintain its labels.

```python
df.loc[['China'], ['Bronze']]
```

As describe in 1.2, for this one, both dimension remain, so we're left with (1, 1), which is a `dataframe`.  

```python
type(df.loc[['China'], ['Bronze']])
```

## Part II: Subsetting, `loc` and `iloc`

### 2.1 Get the intersection of the last two rows and the last two columns **without** use `loc` and `iloc`.


```python
df[-2:][['Bronze', 'Continent']]
```

Do chain indexing, access the left two rows first through slicing, then select the last two column by it's column label

### 2.2 Get the intersection of the last two rows and the last two columns **with at least one** `loc`.

```python
df.loc[['China', 'Thailand']][['Bronze', 'Continent']]
```

As `.loc` only support actual row and column label, we have to select them by it's name, and do chain indexing just like 2.1

### 2.3 Get the intersection of the last two rows and the last two columns **with at least one** `iloc`.

```python
df.iloc[-2:, -2:]
```

Access the last two rows and last two columns by index slicing, as `.iloc` is intended to be used.
