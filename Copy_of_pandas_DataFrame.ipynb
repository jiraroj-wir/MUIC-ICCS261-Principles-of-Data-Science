## What is a DataFrame?
DataFrame is a Data in a table form (like a spreadsheet) where all data points in the same column have the same type.

<img src="https://pynative.com/wp-content/uploads/2021/02/dataframe.png" width=50% />

````python
df = pd.DataFrame({
    'Gold': [15, 12, 10],
    'Silver': [11, 9, 5] ,
    'Bronze': [6, 8, 12],
    'Continent': ['Europe', 'Asia', 'Asia']
    },
    ['France','China','Thailand'])
df
````

## 1. CREATE
The simple way to create a dataframe from scratch is by using a dictionary. A dictionary holds key:value where **key** is column name. The **value** is a **list** of data points.

<img src="https://av-eks-blogoptimized.s3.amazonaws.com/17699Append-a-Dictionary-to-a-list-in-Python-5-i2tutorials.png" />

````python
my_dt = {
    'Gold': [15, 12, 10],
    'Silver': [11, 9, 5] ,
    'Bronze': [6, 8, 12],
    'Continent': ['Europe', 'Asia', 'Asia']
}
df1 = pd.DataFrame(my_dt)
df1
````

Notice that the index of the row is a number. This is called **the label index**. As the word **label** may hint us already, we can set **the label index** to string as well.

````python
my_list = ['France', 'China', 'Thailand']
df2 = pd.DataFrame(my_dt, my_list)
df2
````

Create a dataframe by manually setting each column with a series.

````python
df3 = pd.DataFrame()              # Start with empty dataframe
ser1 = pd.Series([15, 11, 6])     # Create a series from a list
ser2 = pd.Series([11, 9, 5])      # Create a series from a list
ser3 = pd.Series([10, 5, 2])      # Create a series from a list
ser4 = pd.Series(["Europe", "Asia", "Asia"])  # Create a series from a list

df3["Gold"] = ser1
df3["Silver"] = ser2
df3["Bronze"] = ser3
df3["Continent"] = ser4

df3
````

````python
my_index = pd.Series(["France", "China", "Thailand"])  # Create a series from a list
# inplace indicates Python to save the result into the caller variable
df33 = df3.set_index(my_index)
df33
````

````python
df3
````

````python
# Create a series from a list
my_index = pd.Series(["France", "China", "Thailand"])
# inplace indicates Python to save the result into the caller variable
# df3 = df3.set_index(my_index)
df3.set_index(my_index, inplace=True)
df3
````

### 2.3 Both rows and columns
As we just learn that we can use ```iloc[ ]``` to subset data.

We can apply the same method to subset both rows and columns.

Normally, we use indexes to do slicing.

**```df.iloc[row_slicing, column_slicing]```**

````python
df2.iloc[:, 0:1]  # all rows, first column
````

````python
df2.iloc[:, 0]
````

**Prompt**: What is the difference between ```df2.iloc[:, 0:1]``` and ```df2.iloc[:, 0]```?



````python
print(type(df2.iloc[:, 0:1]), type(df2.iloc[:, 0]))
````

````python
df2.iloc[:, 1:]  # all rows, second column to the end
````

````python
df2.iloc[:, 1:-1]  # all rows, second column to the penultimate column
````

````python
df2.iloc[:2, :]  # first two rows, all columns
````

````python
df2.iloc[:2, [0, 3]]  # first two rows, only the Gold and Continent
````

### 4.1 Delete rows
We can use drop( label ) to delete the row.
Note: drop will return a new dataframe.

````python
df3 = df2.drop("China")
df3
````

````python
df2
````

If we want drop to happen to that dataframe, we can set inplace to True.

````python
df4 = df2.copy()  # Let's preserve df2 because we are going to delete right into the dataframe.
df4
````

````python
df4.drop("China", inplace=True)  # Which axis?
df4
````

````python
df4_2 = df2.copy()  # Let's preserve df2 because we are going to delete right into the dataframe.
df4_2 = df4_2.drop("China")
df4_2
````

````python
df5 = df2.drop("Thailand", axis=0)  # row or column?
df5
````

````python
df2  # should remain the same
````

### 4.2 Delete columns
We can use the same <code>drop( )</code>. This time, we just have to set axis to 1.

(the default is <code>axis=0</code> which means row)

````python
df6 = df2.copy()
df6
````

````python
df6.drop("Gold", axis=1, inplace=True)
df6
````

Can we skip the axis argument here?

````python
df7 = df2.copy()
df7.drop("France", inplace=True)
````
