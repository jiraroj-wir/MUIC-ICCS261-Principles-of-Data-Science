**In my own words, what is the difference between a Python list and a NumPy `ndarray`?**

A Python list stores references to Python objects in a dynamic array, so appending is usually cheap, and those objects can have different data types. A numeric NumPy ndarray stores values in a buffer, using a single data type (dtype). It uses shape and strides to locate each entry inside the buffer. This representation supports efficient vectorized calculations. Using Python loops often reduces that performance advantage, so we should avoid it.

**For a 2D array, what does `axis=0` refer to? What does `axis=1` refer to?**

View the ndarray as if it was in a cartesian coordinate system, 

$$A(i, j, k, ..., n)$$, 

then `axis=0` written in `i` would be the outer most shell, while `axis=n` would be the inner most shell. For example

```python
 A = np.array([
      [[1, 2, 3],
       [4, 5, 6]],

      [[10, 20, 30],
       [40, 50, 60]]])
```

could be written as $A(2, 2, 3)$, then **reduction** along an axis would means having every other axis remain constant while doing the operation on that axis, thus the array is reduced, 1 dimension less, into an array of those other axis that remain constant. 

So, `axis=0` would means, the row and column remain stationary, and we do operation on the external layer, as $(1, 10), (2, 20), (3, 30), (4, 40), (5, 50), (6, 60)$, and we're left with $A'(2, 3)$

For `axis=1`, the external layesr and the columns remain, and we do operation on the rows, as $(1, 4), (2, 5), (3, 6), (10, 40), (20, 50), (30, 60)$, and we're left with $A'(2, 3)$

For `axis=2`, the external layers and the rows remain, and we do operation on the columns, as $(1, 2, 3), (4, 5, 6), (10, 20, 30), (40, 50, 60)$, and we're left with $A'(2, 2)$

> There are several common variants of the array standard deviation calculation. Assuming
> the input `a` is a one-dimensional NumPy array and `mean` is either provided as an argument
> or computed as `a.mean()`, NumPy computes the standard deviation of an array as:
>
> ```python
> N = len(a)
> d2 = abs(a - mean)**2  # abs is for complex `a`
> var = d2.sum() / (N - ddof)  # note use of `ddof`
> std = var**0.5
> ```
>
> Different values of the argument `ddof` are useful in different contexts. NumPy’s default
> `ddof=0` corresponds with the expression:
>
> $$\sqrt{\frac{\sum_i |a_i-\bar{a}|^2}{N}}$$
>
> which is sometimes called the “population standard deviation” in the field of statistics
> because it applies the definition of standard deviation to `a` as if `a` were a complete
> population of possible observations.
>
> Many other libraries define the standard deviation of an array differently, e.g.:
>
> $$\sqrt{\frac{\sum_i |a_i-\bar{a}|^2}{N-1}}$$
>
> In statistics, the resulting quantity is sometimes called the “sample standard deviation”
> because if `a` is a random sample from a larger population, this calculation provides the
> square root of an unbiased estimate of the variance of the population. The use of $N-1$ in
> the denominator is often called “Bessel’s correction” because it corrects for bias (toward
> lower values) in the variance estimate introduced when the sample mean of `a` is used in
> place of the true mean of the population. The resulting estimate of the standard deviation is
> still biased, but less than it would have been without the correction. For this quantity, use
> `ddof=1`.
>
> Note that, for complex numbers, `std` takes the absolute value before squaring, so that the

>
